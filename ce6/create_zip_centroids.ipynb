```python
from shapely.geometry import Point, Polygon
import pandas as pd
import geopandas as gpd
import numpy as np
```

```python
zipcode_dataset = pd.read_csv(r"C:\Users\cquering\Documents\Repository\408\ce6\2025_Gaz_zcta_national.txt", sep = '|')
```

```python
zipcodes = []
point_geometries = []
zipcode_centroids_columns = {'zipcode': zipcodes, 'geometry': point_geometries}

for index, row in zipcode_dataset.iterrows():
    zipcodes.append(str(row[('GEOID')]).zfill(5))
    point_geometries.append(Point(row['INTPTLAT'],row['INTPTLONG']))

zipcode_centroids = gpd.GeoDataFrame(zipcode_centroids_columns, crs="EPSG:4326")

zipcode_centroids.head()
#zipcode_centroids.to_file("zipcode_centroids")
```

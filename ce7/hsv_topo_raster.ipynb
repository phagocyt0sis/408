```python
import xarray as xr
import rioxarray
```

```python
hsv_topo = rioxarray.open_rasterio(r"C:\Users\cquering\Documents\Repository\408\ce7\N34W087.SRTMGL1.hgt\N34W087.hgt")
hsv_topo_meters = hsv_topo.rio.reproject("EPSG:3857") #reprojecting to Web Mercator
pixel_size = (hsv_topo_meters.rio.resolution()[0])**2 #calculating pixel size; first value in pixel resolution tuple squared
greater_500 = hsv_topo_meters.where(hsv_topo_meters > 500) #creating data array with just pixels with value over 500
count_greater_500 = greater_500.count().item() #counting pixels with a non-missing value
area_greater_500 = count_greater_500 * pixel_size #calculating total area using pixel size and count over 500
```

```python
print("There are", count_greater_500, "pixels with elevation greater than 500 meters.", "\nPixel size is approximately ", round(pixel_size), "square meters.", "\nThe total area with elevation over 500 meters is approximately ", round(area_greater_500), "square meters.")
```

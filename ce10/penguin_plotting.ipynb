```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
fig = px.histogram(penguin_dataset, x="bill_length_mm")
fig.show()
```

```python
#creating two separate DataFrames for males and females
sex = penguin_dataset['sex']
species = penguin_dataset['species']
body_mass_g = penguin_dataset['body_mass_g']
f_species = []
f_body_mass_g = []
m_species = []
m_body_mass_g = []

for index, row in penguin_dataset.iterrows():
    if sex.iloc[index] == "female":
        f_species.append(species.iloc[index])
        f_body_mass_g.append(body_mass_g.iloc[index])
    elif sex.iloc[index] == "male":
        m_species.append(species.iloc[index])
        m_body_mass_g.append(body_mass_g.iloc[index])
    else:
        continue

females = pd.DataFrame({'Species' : f_species, 'body_mass_g': f_body_mass_g})
males = pd.DataFrame({'Species': m_species, 'body_mass_g': m_body_mass_g})

```

```python
#creating 2 subplots to contrast weight distribution between the sexes
fig, axes = plt.subplots(1,2)

ax1 = axes[0]
ax2 = axes[1]

sns.histplot(
    males, x="body_mass_g", ax=ax1
)
sns.histplot(
    females, x="body_mass_g", ax=ax2
)

fig.show()
```

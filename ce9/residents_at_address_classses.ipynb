```python
import shapely
```

```python
class Residence:
    """
    Attributes
    ----------
    address: str
    latitude: float
    longitude: float
    residents: list

    Methods
    -------
    get_contact_information
    """
    def __init__(self, address, latitude, longitude, residents):
        self.address = address
        self.coords = shapely.Point(latitude, longitude)
        self.residents = residents
    
    def get_contact_information(self):
        contacts = []
        for i in range(len(self.residents)):
            #print(self.residents[i])
            contacts.append((self.residents[i].name, self.residents[i].phone))
        print(contacts)


class Person:
    """
    Attributes
    ----------
    name: str
    phone: str
    address_instance: Residence
    """
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
        

```
